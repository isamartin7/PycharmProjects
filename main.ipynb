### Classes

Employee:
*     Name
*     Email address
*     ID

```python
class Employee:
    def __init__(self, name, email, id):
        self.name = name
        self.email = email
        self.id = id

    def __str__(self):
        return self.name + " " + self.email + " " + str(self.id)

    def print_id(self,word='ff'):
        print(self.id)
```

```python
employee_1 = Employee("Robert","bob23@gmail.com",2046303)
```

```python
print(employee_1.name)
print(employee_1.email)
print(employee_1)
```

### Threshold using OpenCV

```python
img = cv2.imread('/Users/Isa/Documents/Curso Medical Image Processing/data/ml/rotatechest.png',cv2.IMREAD_GRAYSCALE)

threshold_img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)[1]
cv2.imshow('Original Image', img)
cv2.imshow('Thresholded Image', threshold_img)
cv2.waitKey(0)
```

### Edge detection using OpenCV

```python
img = cv2.imread('/Users/Isa/Documents/Curso Medical Image Processing/data/ml/other_op.png',cv2.IMREAD_GRAYSCALE)

edges = cv2.Canny(img,20,100)
cv2.imshow('Edge Image', edges)
cv2.waitKey(0)
```
